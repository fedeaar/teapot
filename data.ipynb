```python
import pandas as pd
# pd. set_option('display.max_rows', None)
```

```python
df = pd.read_csv("labels.csv", names=["original_filename", "label"])
df 
```

```python
df2 = pd.read_csv("images_rows.csv")
df2[["original_filename", "verdict"]]
```

```python
df3 = pd.merge(df, df2, on='original_filename')
df3 = df3[["original_filename", "label", "verdict"]]
df3["predicted"] = df3[["verdict"]].apply(lambda x: [0.0 if y in ["needs_review"] else 2.0 if y in ["compliant"] else 0.0 for y in x])
```

```python
df4 = df3.copy()
df4 = df4[(df4["predicted"] == 0.0) | (df4["predicted"] == 2.0)]
df4 = df4[(df4["label"] == 0.0) | (df4["label"] == 2.0)]
df4
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

y_true = df4['label']
y_pred = df4['predicted']

print(len(y_true))

# Individual metrics
accuracy  = accuracy_score(y_true, y_pred)

# Full report (includes F1-score + per-class breakdown)
print(classification_report(y_true, y_pred))
```
